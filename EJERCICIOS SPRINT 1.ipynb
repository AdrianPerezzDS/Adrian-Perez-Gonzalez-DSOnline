```python
# EJERCICIO 2.2

suma = 0

for numero in range(1, 21):

    
    if numero % 2 == 0:
        continue

    
    if numero % 7 == 0:
        print(f"Encontrado múltiplo de 7: {numero}")
        break

    
    suma += numero
    print(f"Número: {numero}, suma acumulada: {suma}")
```
